```python
# for loop for printing fizz, buzz, and fizzbuzz
for x in range(1, 101):
    if x % 15 == 0: # Divisible by 5 and 3
        print(x, "fizzbuzz")
    elif x % 3 == 0: # Divisible by 3
        print(x, "fizz")
    elif x % 5 == 0: # Divisible by 5
        print(x, "buzz")
    else:
        print(x) # Prints number if it is not divisible by 5 or 3
```
